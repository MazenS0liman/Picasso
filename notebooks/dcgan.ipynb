```python
import numpy as np
import matplotlib.pyplot as plt
import torchvision.utils as vutils
from IPython.display import Image as IPImage, display

%matplotlib inline

from picasso.trainers.dcgan_trainer import DCGANTrainer, DCGANHparams, device
print("Device:", device)
```

## Configure and build the trainer
Leave `datapath=None` to auto-find `datasets/Event/Event`. Set `load_from_checkpoint=True` to resume.

```python
hparams = vars(DCGANHparams(image_size=64, nz=100, ngf=64, ndf=64, lr=0.0002, beta1=0.5))

trainer = DCGANTrainer(
    hparams=hparams,
    datapath="./datasets/Event/Event",
    batch_size=32,
    ckpt_path="checkpoints/dcgan_ckpt.pt",
    sample_dir="samples",
    load_from_checkpoint=False,
)
print(trainer.netG)
print(trainer.netD)
```

## Training images

```python
real_batch = next(iter(trainer.dataloader))
plt.figure(figsize=(8, 8))
plt.axis("off")
plt.title("Training Images")
plt.imshow(np.transpose(vutils.make_grid(real_batch[0][:64], padding=2, normalize=True), (1, 2, 0)))
plt.show()
```

## Train
A checkpoint and a sample grid are saved after every epoch. Interrupting the cell (stop button) still checkpoints.

```python
G_epoch_losses, D_epoch_losses = trainer.fit(epochs=50)
```

## Losses

```python
plt.figure(figsize=(10, 5))
plt.title("Generator and Discriminator Loss During Training")
plt.plot(trainer.G_losses, label="G")
plt.plot(trainer.D_losses, label="D")
plt.xlabel("iterations")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

## Real vs. generated

```python
fake = trainer.generate(64)

plt.figure(figsize=(15, 15))
plt.subplot(1, 2, 1)
plt.axis("off")
plt.title("Real Images")
plt.imshow(np.transpose(vutils.make_grid(real_batch[0][:64], padding=5, normalize=True), (1, 2, 0)))

plt.subplot(1, 2, 2)
plt.axis("off")
plt.title("Fake Images")
plt.imshow(np.transpose(vutils.make_grid(fake, padding=2, normalize=True), (1, 2, 0)))
plt.show()
```
